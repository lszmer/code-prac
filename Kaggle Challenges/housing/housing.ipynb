```python
import pandas as pd
import seaborn as sns
import numpy as np
import statsmodels.api as sm
from patsy import dmatrices
```

```python
df = pd.read_csv('us_house_Sales_data.csv')
```

```python
df
df1 = df.copy()
```

```python
df1['Area'] = df['Area (Sqft)'].str.replace(" sqft", "")
df1['Lot'] = df['Lot Size'].str.replace(" sqft" , "")
df1["Price"] = df["Price"].str.replace(r'[$,]', '', regex=True)
df1["Bedrooms"] = df["Bedrooms"].str.replace(" bds", "")
df1["Bathrooms"] = df["Bathrooms"].str.replace(" ba", "")
```

```python
df1 = df1.drop(["Address", "State", "MLS ID", "Listing URL"], axis=1)
df1
```

```python
cols = ["Area", "Lot", "Price", "Bedrooms", "Bathrooms"]
df1[cols] = df1[cols].apply(pd.to_numeric, errors="coerce")
df1 = df1.drop(["Area (Sqft)", "Lot Size"], axis=1)
```

```python
sns.pairplot(df1, diag_kind="kde", kind="reg")
```

```python
sns.jointplot(data=df1, x="Price", y="Bedrooms")
```
