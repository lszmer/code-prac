# Prepare design matrix and build interaction terms

```python
y = df["electricity_cost"]
X = df.drop("electricity_cost", axis=1)
X = sm.add_constant(X)

dummies = pd.get_dummies(X["structure_type"], prefix = "structure", drop_first=True, dtype=int)
#dummies_interaction = pd.get_dummies(X["structure_type"], drop_first=True, prefix="structure_inter", dtype=int)
#for col in dummies_interaction:
#    X[col] = dummies_interaction[col] * X["resident_count"]
X = pd.concat([X, dummies], axis=1).drop(["structure_type"], axis=1)
```

```python
#standardize = list(numerics)  + dummies_interaction.drop("structure_inter_Industrial", axis=1).columns.to_list()
standardize = list(numerics)
X[standardize]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train[standardize] = scaler.fit_transform(X_train[standardize])
X_test[standardize] = scaler.transform(X_test[standardize])
```

```python
sns.heatmap(X_train.corr(), cmap="coolwarm", annot=True)
```

```python
# OLS
OLS = sm.OLS(y_train, X_train).fit()
print(OLS.summary())

from sklearn.metrics import mean_absolute_error as MAE, mean_squared_error as MSE, r2_score as R2

OLS_preds = OLS.predict(X_test)
print(f"MSE: {MSE(y_test, OLS_preds)}, MAE: {MAE(y_test, OLS_preds)}, R2: {R2(y_test, OLS_preds)}")
```

```python
from sklearn.model_selection import KFold
def kcv_regularized_ols(X, y, alphas, L1_wts, n_splits=5):
    kf = KFold(n_splits=n_splits, shuffle=True, random_state=42)
    results = []

    for alpha in alphas:
        for L1_wt in L1_wts:
            mse_scores = []
            for train_index, test_index in kf.split(X):
                X_train, X_test = X.iloc[train_index], X.iloc[test_index]
                y_train, y_test = y.iloc[train_index], y.iloc[test_index]
                model = sm.OLS(y_train, X_train).fit_regularized(alpha=alpha, L1_wt=L1_wt, refit=True)
                preds = model.predict(X_test)
                mse_scores.append(MSE(y_test, preds))
            results.append({"alpha": alpha, "L1_wt": L1_wt, "MSE": np.mean(mse_scores)})

    return pd.DataFrame(results)
alphas = np.logspace(-4, 1, 10)
L1_wts = np.linspace(0, 1, 5)
kcv_results = kcv_regularized_ols(X_train, y_train, alphas, L1_wts)
print(kcv_results)
```
