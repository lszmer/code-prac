```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

train = pd.read_csv('train.csv')
train.iloc[:, 1:5]
train.head()
```

```python
# Numerical features
train['Transported'] = train['Transported'].astype(int)
numerical_features = train.select_dtypes(include=[np.number]).columns.tolist()

# Impute median for missing values for numerical columns
for col in train[numerical_features]:
    train[col].fillna(train[col].median(), inplace=True)

# Impute mode for missing values for categorical columns
categorical_features = train.select_dtypes(include=[object]).columns.tolist()
for col in train[categorical_features]:
    train[col].fillna(train[col].mode()[0], inplace=True)
# Check for remaining missing values
print(train.isna().sum())
```

```python
# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# More comprehensive EDA for Spaceship Titanic
print("="*50)
print("ENHANCED EXPLORATORY DATA ANALYSIS")
print("="*50)

# First, make sure we have the dataset loaded correctly
# This should match how the original notebook loads the data
train_df = pd.read_csv('train.csv')

# 1. Check class balance for the target variable
print("\n1. Target Variable Distribution:")
target_counts = train_df['Transported'].value_counts(normalize=True) * 100
# Fix: Use dictionary-like access for boolean indices
print(f"False: {target_counts.loc[False]:.2f}%")
print(f"True: {target_counts.loc[True]:.2f}%")

# 2. Analyze categorical variables
print("\n2. Categorical Variables Analysis:")
categorical_cols = ['HomePlanet', 'CryoSleep', 'Destination', 'VIP']

for col in categorical_cols:
    print(f"\n{col} Distribution:")
    value_counts = train_df[col].value_counts()
    print(value_counts)
    
    # Create a figure with two subplots
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))
    
    # Plot overall distribution
    sns.countplot(x=col, data=train_df, ax=ax1)
    ax1.set_title(f'Distribution of {col}')
    ax1.set_ylabel('Count')
    ax1.tick_params(axis='x', rotation=45)
    
    # Plot distribution by target
    sns.countplot(x=col, hue='Transported', data=train_df, ax=ax2)
    ax2.set_title(f'Distribution of {col} by Transport Status')
    ax2.set_ylabel('Count')
    ax2.tick_params(axis='x', rotation=45)
    
    plt.tight_layout()
    plt.show()
    
    # Calculate transport rate by category
    print(f"\nTransport Rate by {col} Category:")
    transport_rate = train_df.groupby(col)['Transported'].mean().sort_values(ascending=False) * 100
    print(transport_rate)

# Process Cabin information for analysis
train_df[["Deck", "Cabin_num", "Side"]] = train_df["Cabin"].str.split("/", expand=True)

# Analyze Deck and Side
for col in ['Deck', 'Side']:
    print(f"\n{col} Distribution:")
    value_counts = train_df[col].value_counts()
    print(value_counts)
    
    # Create a figure with two subplots
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))
    
    # Plot overall distribution
    sns.countplot(x=col, data=train_df, ax=ax1)
    ax1.set_title(f'Distribution of {col}')
    ax1.set_ylabel('Count')
    ax1.tick_params(axis='x', rotation=45)
    
    # Plot distribution by target
    sns.countplot(x=col, hue='Transported', data=train_df, ax=ax2)
    ax2.set_title(f'Distribution of {col} by Transport Status')
    ax2.set_ylabel('Count')
    ax2.tick_params(axis='x', rotation=45)
    
    plt.tight_layout()
    plt.show()

# 3. Analyze numerical variables
print("\n3. Numerical Variables Analysis:")
numerical_cols = ['Age', 'RoomService', 'FoodCourt', 'ShoppingMall', 'Spa', 'VRDeck']

# Summary statistics for numerical variables
print("\nSummary Statistics for Numerical Variables:")
print(train_df[numerical_cols].describe())

# Distribution and relationship with target for each numerical variable
for col in numerical_cols:
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))
    
    # Distribution plot
    sns.histplot(data=train_df, x=col, hue='Transported', kde=True, ax=ax1)
    ax1.set_title(f'Distribution of {col} by Transport Status')
    
    # Box plot
    sns.boxplot(x='Transported', y=col, data=train_df, ax=ax2)
    ax2.set_title(f'Box Plot of {col} by Transport Status')
    
    plt.tight_layout()
    plt.show()

# 4. Correlation analysis
print("\n4. Correlation Analysis:")
# Need to convert boolean columns to numeric for correlation analysis
corr_df = train_df.copy()
# Convert Transported to numeric
corr_df['Transported'] = corr_df['Transported'].astype(int)
# Convert other boolean columns if needed
for col in ['VIP', 'CryoSleep']:
    if corr_df[col].dtype == bool:  # Fix: Check if dtype is bool
        corr_df[col] = corr_df[col].astype(int)

# Select only numeric columns for correlation
numeric_cols = corr_df.select_dtypes(include=['int64', 'float64']).columns
correlation = corr_df[numeric_cols].corr()

# Plot correlation heatmap
plt.figure(figsize=(12, 10))
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.show()

# 5. Missing values analysis
print("\n5. Missing Values Analysis:")
missing_data = train_df.isnull().sum().sort_values(ascending=False)
missing_percent = (train_df.isnull().sum() / train_df.isnull().count() * 100).sort_values(ascending=False)
missing_df = pd.concat([missing_data, missing_percent], axis=1, keys=['Total', 'Percent'])
print(missing_df[missing_df['Total'] > 0])

# Visualize missing values
plt.figure(figsize=(10, 6))
sns.heatmap(train_df.isnull(), yticklabels=False, cbar=False, cmap='viridis')
plt.title('Missing Value Heatmap')
plt.tight_layout()
plt.show()

# 6. Spending patterns analysis
print("\n6. Spending Patterns Analysis:")
# Create total spending column
train_df['TotalSpending'] = train_df['RoomService'] + train_df['FoodCourt'] + train_df['ShoppingMall'] + train_df['Spa'] + train_df['VRDeck']

# Visualize total spending by transport status
plt.figure(figsize=(10, 6))
sns.histplot(data=train_df, x='TotalSpending', hue='Transported', kde=True)
plt.title('Total Spending Distribution by Transport Status')
plt.xlabel('Total Spending')
plt.tight_layout()
plt.show()

# Spending by VIP status
plt.figure(figsize=(10, 6))
sns.boxplot(x='VIP', y='TotalSpending', hue='Transported', data=train_df)
plt.title('Total Spending by VIP Status and Transport Status')
plt.tight_layout()
plt.show()

# 7. Group analysis
print("\n7. Group Analysis:")
# Extract group ID from PassengerId
train_df['GroupId'] = train_df['PassengerId'].str.split('_').str[0]

# Group size distribution
group_sizes = train_df.groupby('GroupId').size().reset_index(name='GroupSize')
train_df = train_df.merge(group_sizes, on='GroupId', how='left')

plt.figure(figsize=(10, 6))
sns.countplot(x='GroupSize', data=train_df)
plt.title('Distribution of Group Sizes')
plt.tight_layout()
plt.show()

# Transport rate by group size
plt.figure(figsize=(10, 6))
sns.barplot(x='GroupSize', y='Transported', data=train_df)
plt.title('Transport Rate by Group Size')
plt.ylabel('Transport Rate')
plt.tight_layout()
plt.show()

# 8. Age distribution by key factors
print("\n8. Age Distribution by Key Factors:")
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

sns.boxplot(x='HomePlanet', y='Age', data=train_df, ax=axes[0, 0])
axes[0, 0].set_title('Age Distribution by Home Planet')

sns.boxplot(x='Destination', y='Age', data=train_df, ax=axes[0, 1])
axes[0, 1].set_title('Age Distribution by Destination')

sns.boxplot(x='CryoSleep', y='Age', data=train_df, ax=axes[1, 0])
axes[1, 0].set_title('Age Distribution by CryoSleep Status')

sns.boxplot(x='VIP', y='Age', data=train_df, ax=axes[1, 1])
axes[1, 1].set_title('Age Distribution by VIP Status')

plt.tight_layout()
plt.show()

# 9. Bivariate analysis of important features
print("\n9. Bivariate Analysis of Important Features:")
# CryoSleep vs Destination with Transport outcome
g = sns.catplot(x='Destination', hue='Transported', col='CryoSleep', 
            data=train_df, kind='count', height=5, aspect=1.2)
g.fig.suptitle('Transport Status by Destination and CryoSleep', y=1.05)
plt.tight_layout()
plt.show()

# HomePlanet vs VIP with Transport outcome
g = sns.catplot(x='HomePlanet', hue='Transported', col='VIP', 
            data=train_df, kind='count', height=5, aspect=1.2)
g.fig.suptitle('Transport Status by HomePlanet and VIP Status', y=1.05)
plt.tight_layout()
plt.show()

# 10. Missing values by transport status
print("\n10. Missing Values by Transport Status:")
# Create a function to analyze missing values by target
def missing_by_target(df, col, target='Transported'):
    # Fix: Create a Series with boolean values for missing data
    missing_series = df[col].isnull()
    # Create a categorical variable for missing status
    df_temp = df.copy()
    df_temp['Missing'] = missing_series.map({True: 'Missing', False: 'Not Missing'})
    
    # Calculate percentages
    crosstab = pd.crosstab(df_temp['Missing'], df_temp[target], normalize='index') * 100
    print(f"\nPercentage of {target} by {col} Missing Status:")
    print(crosstab)
    
    plt.figure(figsize=(8, 5))
    sns.countplot(x='Missing', hue=target, data=df_temp)
    plt.title(f'Count of {target} by {col} Missing Status')
    plt.tight_layout()
    plt.show()

# Analyze missing values for important columns
for col in ['HomePlanet', 'CryoSleep', 'Cabin', 'Destination', 'Age', 'VIP']:
    missing_by_target(train_df, col)

print("\n==== EDA Complete ====")
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    train.drop(columns=['Transported']),
    train['Transported'],
    test_size=0.2,
    random_state=42
)

num_features = X_train.select_dtypes(include='number').columns.tolist()
cat_features = X_train.select_dtypes(include='object').columns.tolist()
bool_features = X_train.select_dtypes(include='bool').columns.tolist()

print(f"Numeric features: {num_features}")
print(f"Categorical features: {cat_features}")
print(f"Boolean features: {bool_features}")

preprocessor = ColumnTransformer([
    ('num', StandardScaler(), num_features),
    ('cat', OneHotEncoder(handle_unknown='ignore'), cat_features)
])
```

# Models


```python
from sklearn.pipeline import Pipeline
from sklearn.model_selection import GridSearchCV, train_test_split

from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, roc_auc_score, roc_curve, auc
```

## CART

```python
pipeline_CART = Pipeline([
    ('preprocessing', preprocessor),
    ('classifier', DecisionTreeClassifier(random_state=42))
])

param_grid = {
    'classifier__max_depth': [3, 5, 10, 15],
    'classifier__criterion': ['gini', 'entropy'],
    'classifier__min_samples_split': [2, 3, 4, 5],
    'classifier__min_samples_leaf': [1, 3, 5, 2, 8],
    'classifier__max_features': [None, 'sqrt', 'log2']
}

grid_search_CART = GridSearchCV(
    pipeline_CART,
    param_grid,
    cv=3,  # 5-fold cross-validation
    scoring='roc_auc',  # or 'f1', 'roc_auc', etc.
    n_jobs=-1,  # use all CPU cores
    refit=True,  # refit the model with the best parameters
    verbose=2
)

grid_search_CART.fit(X_train, y_train)
```

```python
from sklearn.tree import plot_tree
best_tree = grid_search_CART.best_estimator_.named_steps['classifier']
preprocessor = grid_search_CART.best_estimator_.named_steps['preprocessing']
plt.figure(figsize=(16, 10))
plot_tree(best_tree, feature_names=preprocessor.get_feature_names_out(), class_names=True, filled=True)
plt.show()
```

```python
print("Best parameters:", grid_search_CART.best_params_)

y_pred = grid_search_CART.predict(X_test)
y_proba = grid_search_CART.predict_proba(X_test)[:, 1]

print(classification_report(y_test, y_pred))
print("Test AUC:", roc_auc_score(y_test, y_proba))

print("Best cross-val accuracy:", grid_search_CART.best_score_)

# Evaluate on test set
print("Test set accuracy:", accuracy_score(y_test, y_pred))
# ROC Curve for Random Forest
fpr_rf, tpr_rf, thresholds_rf = roc_curve(y_test, grid_search_CART.predict_proba(X_test)[:, 1])
roc_auc_rf = auc(fpr_rf, tpr_rf)
plt.figure(figsize=(8, 6))
plt.plot(fpr_rf, tpr_rf, color='blue', label=f'Random Forest ROC curve (area = {roc_auc_rf:.2f})')
plt.plot([0, 1], [0, 1], color='red', linestyle='--')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Receiver Operating Characteristic (ROC) Curve - Random Forest')
plt.legend(loc='lower right')
plt.show()
```

## XGBoost

```python
print(X_train.columns)
```

```python
from xgboost import XGBClassifier

pipeline_XGB = Pipeline([
    ('preprocessing', preprocessor),
    ('xgb', XGBClassifier(use_label_encoder=False, eval_metric='logloss', random_state=42))
])

param_grid_XGB = {
    'xgb__n_estimators': list(range(50, 500, 100)),  # 50 to 500 with step of 50
    'xgb__max_depth': list(range(1, 12, 3)),
    'xgb__learning_rate': [0.01, 0.1, 0.2],
    'xgb__subsample': [0.7, 1.0]
}

grid_search_XGB = GridSearchCV(
    pipeline_XGB,
    param_grid_XGB,
    cv=5,
    scoring='roc_auc',
    n_jobs=-1,
    verbose=1
)

grid_search_XGB.fit(X_train, y_train)
```

```python
from sklearn.metrics import classification_report, confusion_matrix, roc_curve, auc

print("Best parameters:", grid_search_XGB.best_params_)

# Predict labels and probabilities on the test set using the XGBoost pipeline
y_pred_xgb = grid_search_XGB.predict(X_test)
y_proba_xgb = grid_search_XGB.predict_proba(X_test)[:, 1]  # Probability for the positive class

# Print classification report
print("Classification Report - XGBoost:")
print(classification_report(y_test, y_pred_xgb))

# Confusion matrix
cm_xgb = confusion_matrix(y_test, y_pred_xgb)
print("Confusion Matrix - XGBoost:")
print(cm_xgb)

# Plot ROC Curve
fpr_xgb, tpr_xgb, _ = roc_curve(y_test, y_proba_xgb)
roc_auc_xgb = auc(fpr_xgb, tpr_xgb)

plt.figure(figsize=(8, 6))
plt.plot(fpr_xgb, tpr_xgb, label=f"XGBoost ROC curve (AUC = {roc_auc_xgb:.2f})", color='blue')
plt.plot([0, 1], [0, 1], linestyle='--', color='red')
plt.xlim([0, 1])
plt.ylim([0, 1.05])
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve - XGBoost")
plt.legend(loc="lower right")
plt.show()  
```

```python
# Feature Importance
importances = grid_search_XGB.best_estimator_.named_steps['xgb'].feature_importances_
feature_names = grid_search_XGB.best_estimator_.named_steps['preprocessing'].get_feature_names_out()
# Create a DataFrame for feature importances
feature_importances_df = pd.DataFrame({
    'Feature': feature_names,
    'Importance': importances
})
# Sort by importance
feature_importances_df.sort_values(by='Importance', ascending=False, inplace=True)
# Plot feature importances
plt.figure(figsize=(12, 8))
sns.barplot(x='Importance', y='Feature', data=feature_importances_df.head(20))
plt.title('Top 20 Feature Importances - XGBoost')
plt.xlabel('Importance')
plt.ylabel('Feature')
plt.show()
print(feature_names)
```

```python
# Get the fitted pipeline
fitted_pipeline = grid_search_XGB.best_estimator_

# Get transformed feature matrix (numerical, ready for XGBoost)
X_train_preprocessed = fitted_pipeline.named_steps['preprocessing'].transform(X_train)

# Also get feature names if needed
feature_names = fitted_pipeline.named_steps['preprocessing'].get_feature_names_out()

# Create SHAP explainer
import shap
xgb_model = fitted_pipeline.named_steps['xgb']
explainer = shap.Explainer(xgb_model)

# Compute SHAP values
shap_values = explainer(X_train_preprocessed)

# Summary plot with transformed features
shap.summary_plot(shap_values, features=X_train_preprocessed, feature_names=feature_names)
```

```python
from sklearn.ensemble import RandomForestClassifier

pipeline_RF = Pipeline([
    ('preprocessing', preprocessor),
    ('classifier', RandomForestClassifier(random_state=42))
])

param_grid_RF = {
    'classifier__n_estimators': list(range(10, 201, 50)),  # 10 to 200 with step of 10
    'classifier__criterion': ['gini', 'entropy'],
    'classifier__bootstrap': [True, False],
    'classifier__max_depth': [None, 5, 10],
    'classifier__min_samples_split': [2, 5],
    'classifier__min_samples_leaf': [1, 2],
    'classifier__max_features': ['sqrt', 'log2']
}

grid_search_RF = GridSearchCV(
    pipeline_RF,
    param_grid_RF,
    cv=5,
    scoring='roc_auc',  # or 'accuracy', 'f1', etc.
    n_jobs=-1,
    verbose=2
)

grid_search_RF.fit(X_train, y_train)
```
