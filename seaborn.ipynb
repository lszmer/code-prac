```python
import sklearn as sk
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
crash_df = sns.load_dataset('car_crashes')
```

```python
crash_df.head()
```

## Distribution plot

```python
sns.displot(crash_df['alcohol'], bins=20, kde=True)
plt.title('Alcohol Distribution')
plt.xlabel('Alcohol')
plt.ylabel('Density')
plt.show()
```

## Joint plot

```python
sns.jointplot(x='alcohol', y='speeding', data=crash_df, kind='reg')
# or with kde, hex
```

## Pair plot

```python
sns.pairplot(crash_df, diag_kind='kde', kind='reg')
plt.show()
```

```python
tips_df = sns.load_dataset('tips')
sns.pairplot(tips_df, hue = 'sex', diag_kind='kde', kind='reg')
```

## Rug plot

```python
sns.rugplot(x=tips_df['tip'], height=0.05)
```

```python
sns.set_style('whitegrid')
sns.set_context('paper')
sns.set_palette('pastel')
sns.set(font_scale=1.2)

sns.jointplot(x='size', y='tip', data=tips_df, kind='reg')
```

# Categorical data

```python
sns.barplot(x = 'sex', y = 'total_bill', data=tips_df, estimator=np.cov)
```

```python
sns.countplot(x = 'sex', data=tips_df, palette='pastel')
```

```python
sns.boxplot(x = 'day', y = 'total_bill', data=tips_df, palette='pastel', hue = 'sex')
plt.show()
sns.violinplot(x = 'day', y = 'tip', data=tips_df, palette='pastel', hue = 'sex', split=True)
plt.show()
sns.stripplot(x = 'day', y = 'total_bill', data=tips_df, palette='pastel', hue = 'sex')
plt.show()
```

```python
sns.violinplot(x = 'day', y='total_bill', data = tips_df, split = True, hue = 'sex', palette = 'magma')
plt.legend(loc = 'upper left')
plt.show()

sns.swarmplot(x = 'day', y='total_bill', data = tips_df)
```

# Matrix plots

```python
plt.figure(figsize=(10, 6))


crash_df = sns.load_dataset('car_crashes')
#sns.heatmap(crash_df.corr(), annot=True, cmap='coolwarm')
crash_df.head()
crash_df.drop(['abbrev'], axis=1, inplace=True)

sns.heatmap(crash_df.corr(), annot=True, cmap='coolwarm', fmt='.2f')
```

```python
flights = sns.load_dataset('flights')
flights
flights = flights.pivot_table(index='month', columns='year', values='passengers')
sns.heatmap(flights, annot=False)
```

```python
iris = sns.load_dataset('iris')

species = iris.pop('species')
sns.clustermap(iris)

sns.clustermap(flights, cmap='coolwarm', standard_scale=1)
```

```python
iris = sns.load_dataset('iris')
iris_g = sns.PairGrid(iris, hue='species', palette='husl')
iris_g.map(plt.scatter)
iris_g.map_diag(plt.hist)
iris_g.map_lower(sns.kdeplot)
iris_g.map_upper(sns.boxplot)
```
