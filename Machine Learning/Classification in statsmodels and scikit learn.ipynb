# Storing and exporting a model

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import statsmodels.formula.api as smf
import pickle
import os # For path manipulation

# --- 1. Create a Sample Model (e.g., an OLS model) ---
np.random.seed(42)
data = pd.DataFrame({
    'X': np.random.rand(100) * 10,
    'Z': np.random.rand(100) * 5,
    'Y': 2 + 3 * np.random.rand(100) + 1.5 * np.random.rand(100) + np.random.randn(100) * 2
})
model = smf.ols('Y ~ X + Z', data=data)
results = model.fit()

print("Original model summary:")
print(results.summary())
print("\n" + "="*50 + "\n")

# --- 2. Define File Path ---
# It's good practice to create a 'models' directory or similar
model_dir = 'saved_models'
os.makedirs(model_dir, exist_ok=True) # Create directory if it doesn't exist

# Define the full path for your model file
model_filename = 'my_ols_model.pkl' # Use .pkl or .pickle extension
model_filepath = os.path.join(model_dir, model_filename)

# --- 3. Save the Model ---
# You generally save the 'results' object, not the 'model' object
# The results object contains all the fitted parameters, residuals, etc.
try:
    with open(model_filepath, 'wb') as file:
        pickle.dump(results, file)
    print(f"Model successfully saved to: {model_filepath}")
except Exception as e:
    print(f"Error saving model: {e}")

# --- 4. Load the Model ---
loaded_results = None # Initialize to None
try:
    with open(model_filepath, 'rb') as file:
        loaded_results = pickle.load(file)
    print(f"Model successfully loaded from: {model_filepath}")
except Exception as e:
    print(f"Error loading model: {e}")

# --- 5. Verify the Loaded Model ---
if loaded_results:
    print("\nLoaded model summary:")
    print(loaded_results.summary())

    # You can now use the loaded model for predictions, etc.
    new_data = pd.DataFrame({'X': [5.0, 8.0], 'Z': [2.5, 4.0]})
    predictions = loaded_results.predict(new_data)
    print(f"\nPredictions from loaded model for new data:\n{new_data}\nPredictions: {predictions.values}")

    # Check if original and loaded are identical (should be)
    assert np.allclose(results.params, loaded_results.params), "Parameters do not match!"
    print("\nOriginal and loaded model parameters match.")
```
